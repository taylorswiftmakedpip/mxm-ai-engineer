```python
import pandas as pd

path = "steam_game_reviews.csv"
data = pd.read_csv(path, dtype=object)
data
```

```python
# Определите, зависят ли характеристики `hours_played` и `recommendation` 
# друг от друга в игре Subnautica.


df_sub = data[data["game_name"] == "Subnautica"].copy() 
df_sub.info()
```

```python
df_sub["recommendation"].value_counts()
```

```python
df_sub["recommendation"] = df_sub["recommendation"].apply(
    lambda x: 1 if x == "Recommended" else 0
)
df_sub.head(1)
```

```python
df_sub["recommendation"].value_counts()
```

```python
df_sub["recommendation"].corr(df_sub["hours_played"])
```

```python
df_sub["hours_played"] = df_sub["hours_played"].str.replace(",", "").astype(float)
```

```python
df_sub["hours_played"]
```

```python
df_sub["recommendation"].corr(df_sub["hours_played"])
```

```python
nefative_max = df_sub[df_sub["recommendation"] == 0]["hours_played"].max()
nefative_max
```

```python
positive_min = df_sub[df_sub["recommendation"] == 1]["hours_played"].min()
positive_min
```

```python
# Используя данные из файла `steam_game_reviews.csv`, 
# найдите среднее число дней, прошедших со дня публикации отзывов для
# игры Subnautica.
```

```python
df_sub.head(1)
```

```python
df_sub["date"]
```

```python
df_sub["date"] = pd.to_datetime(df_sub["date"], format="mixed", errors="coerce")
```

```python
df_sub["date"]
```

```python
df_sub["date"].isna().sum()
```

```python
df_sub = df_sub.dropna(subset="date")
df_sub
```

```python
today = pd.Timestamp.now().date()
today
```

```python
df_sub["count_days"] = (pd.to_datetime(today) - df_sub["date"]).dt.days
df_sub["count_days"] 
```

```python
df_sub.sort_values(by="count_days", ascending=False)
```

```python
df_sub["count_days"].mean()
```
