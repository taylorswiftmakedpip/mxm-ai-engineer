## 3. Уникальные значения и частоты

```python
# ЗАДАЧА «Технологополис»
# Условие:
# В городе "Технологополис" ежегодно проводится конференция разработчиков "DevFest". Для участия в
# мероприятии регистрируются сотни разработчиков, и каждый из них оставляет информацию о своем имени,
# уровне опыта (junior, middle, senior) и компании, в которой работает.
#
# Организаторы мероприятия попросили вас как главного аналитика проанализировать данные, чтобы узнать:
#
# - Какие уровни опыта представлены среди участников?
# - Сколько уникальных компаний представлено на конференции?
# - Какое среднее количество рабочих на компанию?
#
# Данные об участниках представлены в виде одномерных списков:
#
# names = ["Анна", "Дмитрий", "Иван", "Анна", "Мария", "Сергей", "Дмитрий"]
# companies = ["TechCorp", "DevSolutions", "CodeFactory", "TechCorp", "WebStudio", "DataDrive", "CodeFactory"]
# experience_levels = ["junior", "middle", "senior", "junior", "middle", "senior", "middle"]

names = ['Анна', 'Дмитрий', 'Иван', 'Анна', 'Мария', 'Сергей', 'Дмитрий']  # Создаёт список имён участников.
companies = ['TechCorp', 'DevSolutions', 'CodeFactory', 'TechCorp',  # Начинает список компаний участников.
             'WebStudio', 'DataDrive', 'CodeFactory']  # Завершает список компаний; порядок соответствует именам.
experience_levels = ['junior', 'middle', 'senior', 'junior', 'middle', 'senior', 'middle']  # Создаёт список уровней опыта участников.

names_series = pd.Series(names)  # Преобразует список имён в Series.
companies_series = pd.Series(companies)  # Преобразует список компаний в Series.
experience_series = pd.Series(experience_levels)  # Преобразует список уровней опыта в Series.
```

```python
unique_experience_levels = experience_series.unique()  # Сохраняет массив уникальных уровней опыта.
unique_experience_levels  # Выведет junior, middle, senior — без повторений.
```

```python
unique_companies_count = companies_series.nunique()  # Считает число уникальных компаний.
unique_companies_count  # Выведет значение 5 — число разных компаний.
```

```python
companies_counts = companies_series.value_counts()  # Считает число участников от каждой компании.
companies_counts  # Выведет компании и количества участников: 2, 2, 1, 1, 1.
```

```python
mean_participants_per_company = companies_counts.mean()  # Вычисляет среднее количество участников на компанию.
mean_participants_per_company  # Выведет значение 1.4 — среднее число участников на компанию.
```

## 6. Преобразование данных и корреляция

```python
# ЗАДАЧА «Тик-так, время плывет»
# Условие:
# Определите, зависят ли характеристики hours_played и recommendation друг от друга в игре Subnautica.

df_Subnautica = data[data['game_name'] == 'Subnautica'].copy()  # Отбирает отзывы Subnautica в отдельную копию таблицы.
df_Subnautica  # Выведет таблицу отзывов Subnautica.
```

```python
df_Subnautica['recommendation'].value_counts()  # Выведет число отзывов каждой категории рекомендации.
```

```python

df_Subnautica['recommendation'] = df_Subnautica['recommendation'].apply(  # Применяет функцию к каждому значению столбца и сохраняет результат.
    lambda x: 1 if x == 'Recommended' else 0  # Заменяет Recommended на 1, другую проверенную категорию — на 0.
)
df_Subnautica.head(3)  # Выведет первые 3 строки текущей таблицы Subnautica.
```

```python
df_Subnautica['recommendation'].value_counts()  # Выведет число отзывов каждой категории рекомендации.
```

```python
df_Subnautica['hours_played'].corr(df_Subnautica['recommendation'])  # Пробует посчитать корреляцию до преобразования строковых часов в числа.
```

```python
df_Subnautica['hours_played'] = (  # Сохраняет преобразованные часы в том же столбце.
    df_Subnautica['hours_played'].str.replace(',', '', regex=False).astype(float)  # Удаляет запятые-разделители тысяч и превращает строки в числа float.
)
df_Subnautica[['hours_played', 'recommendation']].head(3)  # Выведет первые 3 строки двух преобразованных столбцов.
```

```python
correlation = df_Subnautica['hours_played'].corr(df_Subnautica['recommendation'])  # Вычисляет корреляцию Пирсона между часами и числовой рекомендацией.
correlation  # Выведет коэффициент корреляции; близость к нулю означает слабую линейную связь.
```

```python
negative_max = df_Subnautica[df_Subnautica['recommendation'] == 0]['hours_played'].max()  # Находит максимум часов среди отрицательных отзывов.
negative_max  # Выведет значение 397.4 — максимальное число часов среди отрицательных отзывов.
```

```python
positive_min = df_Subnautica[df_Subnautica['recommendation'] == 1]['hours_played'].min()  # Находит минимум часов среди положительных отзывов.
positive_min  # Выведет значение 0.2 — минимальное число часов среди положительных отзывов.
```

## 10. Даты и сортировка

```python
# ЗАДАЧА «Делу время, а потехе - час»
# Условие:
# Используя данные из файла steam_game_reviews.csv, найдите среднее число дней, прошедших со дня публикации отзывов для
# игры Subnautica.

data = pd.read_csv('steam_game_reviews.csv')  # Повторно читает CSV из рабочей папки.
```

```python
df_Subnautica = data[data['game_name'] == 'Subnautica']  # Отбирает отзывы об игре Subnautica.
df_Subnautica  # Выведет отобранную таблицу.
```

```python
df_Subnautica['date']  # Выведет исходные строки дат публикации.
```

```python
df_Subnautica['date'] = pd.to_datetime(  # Преобразует столбец в даты.
    df_Subnautica['date'], format='mixed', errors='coerce'  # Смешанные форматы; ошибки заменяет на NaT.
)
```

```python
df_Subnautica['date']  # Выведет преобразованные даты и NaT для нераспознанных значений.
```

```python
df_Subnautica['date'].isna().sum()  # Выведет число пропусков в столбце дат.
```

```python
df_Subnautica = df_Subnautica.dropna()  # Удаляет строки с пропуском в любом столбце.
```

```python
df_Subnautica['date'].isna().sum()  # Выведет значение 0 — пропусков дат не осталось.
```

```python
today = pd.Timestamp.now().date()  # Получает сегодняшнюю календарную дату.
today  # Выведет дату дня запуска.
```

```python
df_Subnautica['count_days'] = (  # Создаёт столбец с числом прошедших дней.
    pd.to_datetime(today) - df_Subnautica['date']  # Вычитает дату публикации из сегодняшней даты.
).dt.days  # Извлекает число дней из разности дат.
df_Subnautica  # Выведет таблицу с новым столбцом count_days.
```

```python
df_sorted = df_Subnautica.sort_values(by='count_days', ascending=True)  # Сортирует по возрастанию.
df_sorted  # Выведет отзывы от новых к старым.
```

```python
df_Subnautica['count_days'].mean()  # Выведет среднее число дней с публикации отзывов.
```
