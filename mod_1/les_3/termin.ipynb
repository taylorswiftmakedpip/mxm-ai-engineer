## 2. Создание Series и значения

```python
# Создаёт Series из списка; метки автоматически равны 0, 1, 2, 3.
# Выведет Series со значениями 10, 20, 30, 40.
data_list = [10, 20, 30, 40]
series = pd.Series(data_list)
series
```

```python
# Выведет массив значений [10 20 30 40] без меток.
series.values
```

```python
# Выведет numpy.ndarray — тип массива значений этой числовой Series.
type(series.values)
```

## 3. Метки Series

```python
# Создаёт Series: ключи словаря становятся метками, числа — значениями.
# Выведет Series: a → 10, b → 20, c → 30.
data_dict = {'a': 10, 'b': 20, 'c': 30}
series_dict = pd.Series(data_dict)
series_dict
```

```python
# Выведет Index с метками a, b, c. Атрибут index вызывается без скобок.
series_dict.index
```

```python
# Как думаете, будет ли ошибка?
# Заменяет весь индекс; одинаковые метки разрешены.
# Выведет Series: x → 10, x → 20, z → 30.
series_dict.index = ['x', 'x', 'z']
series_dict
```

```python
# Выведет Series с двумя значениями 10 и 20: метка x встречается дважды.
series_dict['x']
```

```python
# Параметр index задаёт метки при создании Series.
# Выведет значения 10, 20, 30, 40, 50 с метками x, y, z, q, w.
series_with_index = pd.Series(
    [10, 20, 30, 40, 50], index=['x', 'y', 'z', 'q', 'w']
)
series_with_index
```

## 4. Создание DataFrame

```python
# Списки хранят имена, компании и уровни опыта семи участников.
# Элементы с одной позицией относятся к одному участнику.
names = ['Анна', 'Дмитрий', 'Иван', 'Анна', 'Мария', 'Сергей', 'Дмитрий']
companies = [
    'TechCorp', 'DevSolutions', 'CodeFactory', 'TechCorp',
    'WebStudio', 'DataDrive', 'CodeFactory'
]
experience_levels = ['junior', 'middle', 'senior', 'junior', 'middle', 'senior', 'middle']
```

```python
# Списки задают значения столбцов, ключи словаря — их названия.
# Выведет таблицу из 7 строк и 3 столбцов с индексом от 0 до 6.
df = pd.DataFrame({
    "Name": names,
    "Company": companies,
    "Experience Level": experience_levels
})
df
```

```python
# Параметр index задаёт семь меток строк.
# Выведет таблицу участников с метками a, b, c, d, e, f, g.
df = pd.DataFrame({
    "Name": names,
    "Company": companies,
    "Experience Level": experience_levels
}, index=['a', 'b', 'c', 'd', 'e', 'f', 'g'])
df
```

## 5. Подготовка Series и set_axis

```python
# Подключает Pandas под именем pd для работы с индексами.
import pandas as pd
```

```python
# Создаёт Series из трёх чисел с буквенными метками.
# Выведет Series: a → 10, b → 20, c → 30.
series = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
series
```

```python
# Назначает новые метки тем же позициям, не меняя значения.
# Выведет Series: x → 10, y → 20, z → 30. Сама series не меняется.
series_new = series.set_axis(['x', 'y', 'z'])
series_new
```

## 6. Перестройка Series

```python
# Выбирает значения по меткам a, b, d; c исключается.
# Метки d не было: fill_value=0 задаёт для неё ноль.
# Выведет Series: a → 10, b → 20, d → 0.
series_reindex = series.reindex(['a', 'b', 'd'], fill_value=0)
series_reindex
```

## 7. Сброс индекса Series

```python
# Превращает старый индекс в столбец; новые метки строк равны 0, 1, 2.
# Выведет тип pandas.core.frame.DataFrame.
series_reset = series.reset_index()
type(series_reset)
```

```python
# Выведет таблицу: столбец index содержит a, b, c, столбец 0 — 10, 20, 30.
series_reset
```

```python
# drop=True отбрасывает старый индекс, не создавая столбец.
# Выведет тип pandas.core.series.Series.
series_reset = series.reset_index(drop=True)
type(series_reset)
```

```python
# Выведет Series со значениями 10, 20, 30 и новым индексом 0, 1, 2.
series_reset
```

```python
# name задаёт имя столбца значений.
# Выведет таблицу со столбцами index (a, b, c) и value (10, 20, 30).
series_reset = series.reset_index(name='value')
series_reset
```

```python
# inplace=True меняет саму series; для этого нужен drop=True.
# Метод возвращает None, а последнее выражение выводит изменённую Series.
# Выведет значения 10, 20, 30 с индексом 0, 1, 2.
series.reset_index(drop=True, inplace=True)
series
```

## 8. Столбец как индекс DataFrame

```python
# Создаёт и выводит таблицу со столбцами A (a, b, c) и B (10, 20, 30).
df = pd.DataFrame({
    'A': ['a', 'b', 'c'],
    'B': [10, 20, 30]
})
df
```

```python
# Назначает столбец A индексом новой таблицы.
# Выведет таблицу со столбцом B и индексом a, b, c; df не меняется.
df_new = df.set_index('A')
df_new
```

```python
# drop=False сохраняет A среди столбцов.
# Выведет таблицу, где A используется и как индекс, и как столбец данных.
# пример параметра drop=False
df_new = df.set_index('A', drop=False)
df_new
```

```python
# append=True добавляет новый уровень к прежнему индексу.
# Выведет столбец B с составными метками (0, a), (1, b), (2, c).
# пример параметра append=True
df_new = df.set_index('A', append=True)
df_new
```

```python
# inplace=True меняет сам df; результат метода не присваивается обратно.
# Выведет таблицу со столбцом B и индексом a, b, c.
# пример параметра inplace=True
df.set_index('A', inplace=True)
df
```

## 9. Сброс индекса DataFrame

```python
# Создаёт и выводит таблицу со столбцом A: x → 10, y → 20, z → 30.
df = pd.DataFrame({
    'A': [10, 20, 30],
}, index=['x', 'y', 'z'])
df
```

```python
# names задаёт имя столбца из прежнего индекса.
# Выведет таблицу: values содержит x, y, z, A содержит 10, 20, 30.
# Новые метки строк — 0, 1, 2.
df_reset = df.reset_index(names='values')
df_reset
```

## 10. Перестройка DataFrame

```python
# Создаёт и выводит таблицу 2 × 2: столбец A — 1, 2, столбец B — 3, 4.
df = pd.DataFrame({
    'A': [1, 2],
    'B': [3, 4]
}, index=['x', 'y'])
df
```

```python
# index и columns задают нужные метки строк и столбцов.
# fill_value=0 заполняет новые места, а не уже существовавшие пропуски.
# Выведет таблицу 3 × 3; новая строка z и новый столбец C содержат нули.
df_reindexed = df.reindex(
    index=['x', 'y', 'z'], columns=['A', 'B', 'C'], fill_value=0
)
df_reindexed
```
