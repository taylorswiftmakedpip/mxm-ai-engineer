## 8. Задача «Расчёт маршрута для велогонки»

```python
# ЗАДАЧА «Расчёт маршрута для велогонки»
# Условие:
# Вы являетесь организатором велогонки, которая будет проходить в горной местности. Гонка состоит из
# нескольких этапов с различными дистанциями, и вам нужно рассчитать следующие важные параметры для
# гонщиков: 1) Длина каждого этапа - гонка начинается с лёгкого этапа длиной 5 км и заканчивается
# сложным 50 км, всего 10 этапов. 2) Самый короткий и самый длинный этапы гонки. 3) Длина всей
# велогонки. 4) Средняя длина этапов гонки. 5) Общая сложность гонки - рассчитывается как
# произведение всех этапов гонки.
#
# Напишите программу для вычисления указанных параметров и их вывода на экран.

stages = np.linspace(5, 50, 10)  # Создаёт десять равномерно расположенных длин этапов от 5 до 50 км.
stages  # Выведет длины этапов: 5, 10, 15, …, 50.
```

```python
min_stage = np.min(stages)  # Находит минимальную длину этапа — 5 км.
max_stage = np.max(stages)  # Находит максимальную длину этапа — 50 км.
print('Самый короткий этап:', min_stage, 'км')  # Выведет минимальную длину этапа — 5.0 км.
print('Самый длинный этап: ', max_stage, 'км')  # Выведет максимальную длину этапа — 50.0 км.
```

```python
total_length = np.sum(stages)  # Суммирует длины этапов; получается 275 км.
print('Общая длина:', total_length, 'км')  # Выведет общую длину гонки — 275.0 км.
```

```python
average_length = np.mean(stages)  # Находит среднюю длину этапа — 27.5 км.
print('Средняя длина этапа:', average_length, 'км')  # Выведет среднюю длину этапа — 27.5 км.
```

```python
total_complexity = np.prod(stages)  # Перемножает длины этапов для условного показателя сложности.
print('Условная сложность:', total_complexity)  # Выведет произведение длин этапов — 35437500000000.0.
```

## 10. Разбор ДЗ «Лотерея»

```python
# ЗАДАЧА «Лотерея»
# Условие:
# Нам предстоит организовать лотерею, где разыгрываются призы среди участников. Каждый билет
# содержит 6 чисел в диапазоне от 4 до 42 включительно. Призы распределяются относительно суммы
# чисел в билете по следующим категориям: 1) Если баллов 50 и меньше, приз – «Утешительный подарок»
# 2) Если баллов от 51 до 100, приз – «Средний выигрыш» 3) Если баллов от 101 и выше, приз –
# «Высокая награда»
#
# Напишите программу для случайной генерации билетов среди 48 участников. На экран выводится номер
# участника, его билет, сумма чисел в билете и полученный приз.

for number in range(48):  # Перебирает 48 участников; в переменной номера от 0 до 47.
    lottery_ticket = np.random.randint(4, 43, 6)  # Создаёт билет из шести целых чисел от 4 до 42 включительно.
```

```python
# Призы распределяются по сумме чисел: до 100, от 101 до 150, от 151.

for number in range(48):  # Перебирает 48 участников; в переменной номера от 0 до 47.
    lottery_ticket = np.random.randint(4, 43, 6)  # Создаёт билет из шести целых чисел от 4 до 42 включительно.
    sum_numbers = np.sum(lottery_ticket)  # Считает сумму чисел в текущем билете.
    if sum_numbers <= 100:  # Сумма до 100 включительно.
        print(number + 1, lottery_ticket, sum_numbers, "Утешительный приз")  # Выведет номер участника, билет, сумму и название приза.
    elif 101 <= sum_numbers <= 150:  # Сумма от 101 до 150 включительно.
        print(number + 1, lottery_ticket, sum_numbers, "Средний выигрыш")  # Выведет номер участника, билет, сумму и название приза.
    else:  # Сумма от 151.
        print(number + 1, lottery_ticket, sum_numbers, "Высокая награда")  # Выведет номер участника, билет, сумму и название приза.
```

```python
# ЗАДАЧА «Лотерея»
# Условие:
# Нам предстоит организовать лотерею, где разыгрываются призы среди участников. Каждый билет
# содержит 6 чисел в диапазоне от 4 до 42 включительно. Призы распределяются относительно суммы
# чисел в билете по следующим категориям: 1) Если баллов 50 и меньше, приз – «Утешительный подарок»
# 2) Если баллов от 51 до 100, приз – «Средний выигрыш» 3) Если баллов от 101 и выше, приз –
# «Высокая награда»
#
# Напишите программу для случайной генерации билетов среди 48 участников. На экран выводится номер
# участника, его билет, сумма чисел в билете и полученный приз.
# Призы распределяются по сумме чисел: до 50, от 51 до 100, от 101.

np.random.seed(42)  # Задаёт начальное состояние генератора случайных чисел.

for number in range(48):  # Перебирает 48 участников; в переменной номера от 0 до 47.
    lottery_ticket = np.random.randint(4, 43, 6)  # Создаёт билет из шести целых чисел от 4 до 42 включительно.
    sum_numbers = np.sum(lottery_ticket)  # Считает сумму чисел в текущем билете.

    if sum_numbers <= 50:  # Сумма до 50 включительно.
        prize = 'Утешительный подарок'  # Сохраняет название приза для соответствующей ветви условия.
    elif sum_numbers <= 100:  # Сумма от 51 до 100 включительно.
        prize = 'Средний выигрыш'  # Сохраняет название приза для соответствующей ветви условия.
    else:  # Сумма от 101.
        prize = 'Высокая награда'  # Сохраняет название приза для соответствующей ветви условия.

    print(f'Участник {number + 1:2}: {lottery_ticket}; сумма {sum_numbers:3}; {prize}')  # Выведет номер участника, билет, сумму и приз по условию задачи.
```

## 11. NumPy и list: память

```python
from sys import getsizeof  # Подключает функцию getsizeof для оценки размеров объектов.
```

```python
list_data = [i for i in range(1000000)]  # Создаёт список из миллиона целых чисел от 0 до 999999.
print('Список без размеров объектов-элементов:', getsizeof(list_data), 'байт')  # Выведет размер самого списка в байтах, без объектов целых чисел.
```

```python
array_data = np.arange(1000000)  # Создаёт массив из миллиона целых чисел от 0 до 999999.
print('Данные массива:', array_data.nbytes, 'байт')  # Выведет объём памяти элементов массива в байтах, без служебной части.
print('Тип элементов:', array_data.dtype)  # Выведет целочисленный тип элементов массива.
print('Байт на элемент:', array_data.itemsize)  # Выведет размер одного элемента в байтах.
```

```python
# Оценка памяти с учётом целых чисел списка и служебной части массива.

list_total_bytes = getsizeof(list_data) + sum(getsizeof(value) for value in list_data)  # Складывает размер списка и размеры его объектов-элементов.
array_total_bytes = getsizeof(array_data)  # Определяет размер владеющего данными массива вместе со служебной частью.

print(f'Список вместе с целыми числами: {list_total_bytes:,} байт')  # Выведет оценку размера в байтах с разделителями разрядов.
print(f'Массив вместе со служебной частью: {array_total_bytes:,} байт')  # Выведет оценку размера в байтах с разделителями разрядов.
```

## 12. NumPy и list: время выполнения

```python
from timeit import timeit  # Подключает функцию timeit для измерения времени.
```

```python
# Замер суммы для списка из 1 000 000 элементов.

def sum_list():  # Определяет функцию для измерения создания данных и вычисления суммы.
    return sum([x for x in range(1000000)])  # Создаёт миллион элементов и возвращает их сумму.

timeit(sum_list, number=100)  # Выведет суммарное время 100 запусков в секундах, а не среднее время одного запуска.
```

```python
# Замер суммы для массива из 10 000 000 элементов.

def sum_array():  # Определяет функцию для измерения создания данных и вычисления суммы.
    return np.sum(np.arange(10000000))  # Создаёт десять миллионов элементов и возвращает их сумму.

timeit(sum_array, number=100)  # Выведет суммарное время 100 запусков в секундах, а не среднее время одного запуска.
```

```python
# Время включает создание списка и поиск минимума.

timeit('min([x for x in range(1000000)])', number=100)  # Выведет суммарное время 100 запусков в секундах, а не среднее время одного запуска.
```

```python
# Замер минимума для массива из 10 000 000 элементов.
# Импорт из setup не входит в измеряемое время.

timeit('np.min(np.arange(10000000))', setup='import numpy as np', number=100)  # Выведет суммарное время 100 запусков в секундах, а не среднее время одного запуска.
```

```python
NUMBER = 100  # Число повторений каждого замера.
```

```python
# Сумма: по 1 000 000 элементов в списке и массиве.

def sum_list():  # Определяет функцию для измерения создания данных и вычисления суммы.
    return sum([x for x in range(1000000)])  # Создаёт миллион элементов и возвращает их сумму.


def sum_array():  # Определяет функцию для измерения создания данных и вычисления суммы.
    return np.sum(np.arange(1000000, dtype=np.int64))  # Создаёт миллион элементов и возвращает их сумму.


list_sum_time = timeit(sum_list, number=NUMBER)  # Сохраняет суммарное время создания списка и вычисления суммы за NUMBER запусков.
array_sum_time = timeit(sum_array, number=NUMBER)  # Сохраняет суммарное время создания массива и вычисления суммы за NUMBER запусков.

print(f'list:  всего {list_sum_time:.4f} с; один запуск {list_sum_time / NUMBER:.6f} с')  # Выведет суммарное время и среднее время одного запуска.
print(f'NumPy: всего {array_sum_time:.4f} с; один запуск {array_sum_time / NUMBER:.6f} с')  # Выведет суммарное время и среднее время одного запуска.
```

```python
# Минимум: по 1 000 000 элементов в списке и массиве.

list_min_time = timeit(  # Измеряет создание списка и поиск минимума NUMBER раз.
    'min([x for x in range(1000000)])',
    number=NUMBER,
)

array_min_time = timeit(  # Измеряет создание массива и поиск минимума NUMBER раз; импорт в setup не измеряется.
    'np.min(np.arange(1000000))',
    setup='import numpy as np',
    number=NUMBER,
)

print(f'list:  всего {list_min_time:.4f} с; один запуск {list_min_time / NUMBER:.6f} с')  # Выведет суммарное время и среднее время одного запуска.
print(f'NumPy: всего {array_min_time:.4f} с; один запуск {array_min_time / NUMBER:.6f} с')  # Выведет суммарное время и среднее время одного запуска.
```
