Нам предстоит организовать лотерею, где разыгрываются призы среди участников. Каждый билет содержит 6 чисел в диапазоне от 4 до 42 включительно.

Призы распределяются относительно суммы чисел в билете по следующим категориям:

1) Если баллов 50 и меньше, приз — «Утешительный подарок»;

2) Если баллов от 51 до 100, приз — «Средний выигрыш»;

3) Если баллов от 101 и выше, приз — «Высокая награда».

Напишите программу для случайной генерации билетов среди 48 участников. На экран выводится номер участника, его билет, сумма чисел в билете и полученный приз.

Какой результат пользователи будут получать чаще всего?

```python
low = mid = high = 0
for user in range(48):
    ticket = np.random.randint(4, 43, 6)
    # print(ticket)
    sum_ticket = np.sum(ticket)
    
    # print(sum_ticket)

    if sum_ticket <100: 
        low +=1
        prize = "Утешительный подарок"
    elif 101 < sum_ticket <= 151:
        mid +=1
        prize = "Средний выигрыш"
    else:
        high +=1
        prize = "Высокая награда"

    print(f"{user} {ticket} {sum_ticket} {prize}")
print(low, mid, high)
```
