```python
data = {
    'species': ['setosa'] * 4 + ['versicolor'] * 4,
    'sepal_length': [5.1, 4.9, 4.7, 4.6, 7.0, 6.4, 6.9, 5.5],
    'petal_length': [1.4, 1.4, 1.3, 1.5, 4.7, 4.5, 4.9, 4.0]
}

# data

# Сгруппируйте данные по species и вычислите среднее значение sepal_length.

# Постройте столбчатую диаграмму (bar) средних значений sepal_length по видам, 
# чтобы наглядно сравнить группы. Подпишите заголовок графика и ось Y.

# Выведите в качестве ответа среднее значение sepal_length для versicolor,
#  округлённое до 2 знаков после запятой.
```

```python
import pandas as pd
df = pd.DataFrame(data)
df
```

```python
grouped = df.groupby("species")["sepal_length"].mean()
grouped
```

```python
answer = grouped["versicolor"]
answer = round(answer, 2)
answer

round(grouped["versicolor"], 2)
# Ответ в лк 6.45
```

```python
grouped.plot(
    kind="bar",
    xlabel="Вид",
    ylabel="Среднее значение длины",
    color=["tab:green", "tab:orange"],
)
```

```python
data = {
    'A': [1.0, 2.0, None, 4.0, 5.0],
    'B': [10.0, None, 30.0, 40.0, 50.0],
    'C': [100.0, 200.0, 300.0, None, 500.0]
}
data
```

```python
df = pd.DataFrame(data)
df
```

```python
df_isna = df.isna().sum()
df_isna
```

```python
df_isna.plot(
    kind="bar",
    xlabel="Столбец",
    ylabel="Количество пропусков",
    #color="#EC431D"
    color=["#49FCED", "purple", "#FFFF00"],
)

# HEX цветовой круг
```

```python
df = df.fillna(df.median())
df2_isna = df.isna().sum()
df2_isna

df2_isna.plot(
    kind="bar",
    xlabel="Столбец",
    ylabel="Количество пропусков",
    color=["#49FCED", "purple", "#FFFF00"],
)

# Ответ в ЛК 0
```
