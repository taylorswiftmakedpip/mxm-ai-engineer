```python
def split_username_and_game_count(value):
    try:
        if "\n" in value:
            parts = value.split("\n")
            username =parts[0]
            game_count = parts[1].split(" ")[0]
        else:
            username = value
            game_count = None
    except:
        print(value)
    return username, game_count

data = data.dropna()
data["game_count"] = data["username"].apply(lambda x: split_username_and_game_count(x)[1]) # game_count
data["username"] = data["username"].apply(lambda x: split_username_and_game_count(x)[0])
```

```python
data.head(1)
```

```python
data.info()
```

```python
bins = [0, 11, 51, 201, 401, float("inf")]
labels = ["новичок", "любитель", "коллекционер", "Гуру", "Повелитель мира игр"]

data["game_count"] = data["game_count"].str.replace(",", "").apply(float)
data["game_category"] = pd.cut(data["game_count"], bins=bins, labels=labels)
data
```

```python
data_clean = data.dropna()
data_clean
data_clean.info()
```

```python
data_clean.head(1)
```

```python
grouped_category = data_clean.groupby(["game_category", "game_name"]).size()
```

```python
type(grouped_category)
```

```python
grouped_category.index
```

```python
grouped_category.index.get_level_values("game_name").isin(["69 Balls", "Counter-Strike 2"])
```

```python
grouped_plays = grouped_category[
    grouped_category.index.get_level_values("game_name").isin(["69 Balls", "Counter-Strike 2"])
]
grouped_plays = grouped_plays.unstack()

# новичок	183
# любитель	346
# коллекционер	722
# Гуру	350
# Повелитель мира игр	519
```
